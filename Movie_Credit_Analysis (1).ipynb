```python
# 1.1Function to extract the first cast member's gender, name, and character

def extract_first_cast_info(cast_data):
    cast_list = json.loads(cast_data)  #
    if cast_list:
        first_cast_member = cast_list[0]
        gender = 'Male' if first_cast_member.get('gender') == 2 else 'Female' if first_cast_member.get('gender') == 1 else 'Other'
        return first_cast_member.get('name'), first_cast_member.get('character'), gender
    return None, None, None

# 1.2Function to extract the first crew member's department, job, and name
def extract_first_crew_info(crew_data):
    crew_list = json.loads(crew_data)
    if crew_list:
        first_crew_member = crew_list[0]
        gender = 'Male' if first_crew_member.get('gender') == 2 else 'Female' if first_crew_member.get('gender') == 1 else 'Other'

        return first_crew_member.get('name'), first_crew_member.get('department'), first_crew_member.get('job'), gender
    return None, None, None


df_credit[['actor_name', 'character_name', 'cast_gender']] = df_credit['cast'].apply(lambda x: pd.Series(extract_first_cast_info(x)))
df_credit[['director_name', 'department', 'job_name', 'crew_gender']] = df_credit['crew'].apply(lambda x: pd.Series(extract_first_crew_info(x)))

important_columns = df_credit[['movie_id', 'title', 'actor_name', 'character_name', 'cast_gender',
                             'director_name', 'department', 'job_name', 'crew_gender']]

important_columns.head()
```

```python
#Joined 2 main dataset for analysis

joined_df = df_credit.join(df_movie.set_index('id'), on='movie_id', how='inner', lsuffix='_credits', rsuffix='_movies')
joined_df.head(5)
```

```python
joined_df.shape
```

```python
actor_profit = joined_df.groupby('actor_name')['profit'].sum()
actor_profit_10 = actor_profit.nlargest(10)
actor_profit_10
```

```python
fig = px.bar(
    x=actor_profit_10.index,
    y=actor_profit_10.values,
    title='Top10 profit actor',
    labels={'x':'Actors', 'y': 'Profit'},
    color=actor_profit_10.values,
    color_continuous_scale='viridis'
)

fig.update_layout(showlegend=False)
fig.show()
```

```python
director_top10 = joined_df.groupby('director_name')['profit'].sum()
director_top10 = director_top10.nlargest(10)
director_top10
```

```python
fig = px.bar(
    x=director_top10.index,
    y=director_top10.values,
    title='Top 10 Director',
    labels={'x':'Director', 'y':'Profit'},
    color=director_top10.values,
    color_continuous_scale='viridis'
)
fig.update_layout(showlegend=False)
fig.show()
```

```python
#Select last 30 years most popular actors

last_30 = joined_df[joined_df['year'] >= 1990]

popular_actors_last30 = last_30.groupby(['year', 'actor_name'])['vote_average'].mean()
popular_actors_last30 = popular_actors_last30.nlargest(10)
popular_actors_last30
```

```python
popular_actors_years_10_df = popular_actors_last30.reset_index()

fig = px.line(
    popular_actors_years_10_df,
    x='actor_name',
    y='year',
    title='Top 10 rating Actor since 1990',
    markers=True,
    hover_data={'actor_name': False, 'year': True, 'vote_average': True}  # Display year and rating on hover

)

fig.update_traces(line_color='#1f77b4', line_width=3)
fig.update_layout(hovermode='x unified')
fig.show()
```

```python
popular_director_years_10 = last_30.groupby(['year', 'director_name'])['vote_average'].mean()
popular_director_years_10 = popular_director_years_10.nlargest(10).reset_index()
popular_director_years_10
```

```python
fig = px.line(
    popular_director_years_10,
    x='director_name',
    y='year',
    title='Top 10 rating Director since 1990',
    markers=True,
    hover_data={'director_name': False, 'year': True, 'vote_average': True}  # Display year and rating on hover

)

fig.update_traces(line_color='#1f77b4', line_width=3)
fig.update_layout(hovermode='x unified')
fig.show()
```

```python
actor_profit_5 = last_30.groupby(['actor_name', 'year'])['profit'].sum()
actor_profit_5 = actor_profit_5.nlargest(5)
actor_profit_5
```

```python
actor_profit_5_df = actor_profit_5.reset_index()

fig = px.line(
    actor_profit_5_df,
    x='year',  # x-axis is the year
    y='profit',  # y-axis is the profit
    color='actor_name',  # Color lines by actor
    markers=True,  # Add markers on the lines
    title='Top 5 Actors by Total Profit Over Years'
)

fig.update_layout(
    xaxis_title='Year',
    yaxis_title='Profit',
    hovermode='x unified'  # Show tooltip for each actor when hovering
)

fig.show()
```
